```python
import pandas as pd

```

```python
index = pd.MultiIndex.from_tuples(
    [
        ('Engineering', 'CSE'),
        ('Engineering', 'ECE'),
        ('Science', 'Physics'),
        ('Science', 'Chemistry')
    ],
    names=['Department', 'Branch']
)

# Create tabular data
data = pd.DataFrame(
    {
        '2025': [85, 78, 92, 88],
        '2026': [90, 82, 95, 91]
    },
    index=index
)
```

```python
print("Original Tabular Data:")
print(data)
```

```python
unstacked_data = data.unstack()

print("\nData after Unstack():")
print(unstacked_data)
```

```python
stacked_data = unstacked_data.stack()

print("\nData after Stack():")
print(stacked_data)
```

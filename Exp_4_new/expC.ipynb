```python
import pandas as pd

```

```python
df1 = pd.DataFrame(
    {
        'Name': ['kanishka', 'Shiva', 'surya'],
        'Marks': [80, None, 98],
        'Grade': ['A', 'B', None]
    },
    index=[101, 102, 103]
)

# Create the second DataFrame
df2 = pd.DataFrame(
    {
        'Name': ['Karthik', 'Likith', 'Abhi'],
        'Marks': [82,76,98],
        'Grade': [None, 'A', 'B']
    },
    index=[101, 102, 104]
)
```

```python
print("First DataFrame:")
print(df1)

print("\nSecond DataFrame:")
print(df2)
```

```python
merged = pd.merge(
    df1,
    df2,
    left_index=True,
    right_index=True,
    how='outer',
    suffixes=('_DF1', '_DF2')
)
```

```python
print("\nMerged DataFrame:")
print(merged)
```

```python
combined = df1.combine_first(df2)

print("\nDataFrame after combine_first():")
print(combined)
```

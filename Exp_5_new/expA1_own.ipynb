```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/penguins.csv"
df=pd.read_csv(url)
df = df[["species", "island", "bill_length_mm","bill_depth_mm"]]
```

```python
print("First 5 rows\n")
print(df.head())
```

```python
print("Information about data: \n")
print(df.info())
```

```python
print("Statistical summary: \n")
print(df.describe())
```

```python
# 1. Matplotlib - Line Plot

plt.figure(figsize=(8, 5))

plt.plot(df.index[:20], df["bill_length_mm"][:20], label="Bill Length")
plt.plot(df.index[:20], df["bill_depth_mm"][:20], label="Bill Depth")

plt.xlabel("Sample Index")
plt.ylabel("Measurement")
plt.title("Bill Length and Bill Depth")

plt.legend()
plt.grid()
plt.show()
```

```python
# 2. Seaborn - Scatter Plot

plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="bill_length_mm",
    y="bill_depth_mm",
    hue="species"
)

plt.title("Bill Length vs Bill Depth")
plt.xlabel("Bill Length")
plt.ylabel("Bill Depth")
plt.show()
```

```python
# 3. Seaborn - Histogram

plt.figure(figsize=(8, 5))

sns.histplot(
    data=df,
    x="bill_length_mm",
    hue="species",
    kde=True
)

plt.title("Distribution of Bill Length")
plt.xlabel("Bill Length")
plt.ylabel("Frequency")
plt.show()
```

```python
# 4. Seaborn - Box Plot

plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x="species",
    y="bill_length_mm"
)

plt.title("Bill Length by Species")
plt.xlabel("Species")
plt.ylabel("Bill Length")
plt.show()
```

```python
# 5. Seaborn - Pair Plot

sns.pairplot(
    df,
    hue="species"
)

plt.show()
```
